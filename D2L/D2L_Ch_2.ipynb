### 2.1.5 Saving Memory

In DL, we typically want memory addresses of parameters to stay fixed since we are constantly updating the parameters every training example/batch. However, if we write smth like Y = X + Y, we dereference the tensor Y use to point at and instead give it a new mmemory address. This is because Python first compute Y + X, and then assigns it a new memory address to the resultant sum. 

Force it to stay in a fixed address by using a previously allocated array Y and tell the new value to go there using slice notation like Y[:] = new_value_or_expression

If the previous value of X is not reused later, we can use the following two options to reduce memory overhead of operation:
1. X[:] = X + Y
2. X += Y

```python
before = id(Y)
Y = Y + X
id(Y) == before
```

```python
Z = torch.zeros_like(Y)
print('id(Z):', id(Z))
Z[:] = X + Y
print('id(Z):', id(Z))
```

```python
before = id(X)
X += Y
id(X) == before
```

### 2.3.9 Matrix-Vector Products

Product between an $m \times n$ matrix $\mathbf{A}$ and an $n$-dimensional vector $\mathbf{x}$.

We can visualize our matrix in terms of its row vectors as follows:

$$\mathbf{A} = \begin{bmatrix} \mathbf{a}_1^\text{T} \\ \mathbf{a}_2^\text{T}  \\ \vdots \\ \mathbf{a}_m^\text{T} \end{bmatrix}$$

where each $\mathbf{a}_i^\text{T} \in \mathbb{R}^n$ is a row vector representing the $i$-th row of the matrix.

The matrix-vector product $\mathbf{Ax}$ is a column vector of length $m$, where each element is the inner (dot) product of the row vector and the vector $\mathbf{x}$:

$$\mathbf{Ax} = \begin{bmatrix} \mathbf{a}_1^\text{T}\mathbf{x} \\ \mathbf{a}_2^\text{T}\mathbf{x} \\ \vdots \\ \mathbf{a}_m^\text{T}\mathbf{x} \end{bmatrix}$$

In PyTorch, this operation can be performed using the torch.mv(A, x) function, OR the @ operator (which supports both matrix-vector and matrix-matrix products).

```python
A.shape, x.shape, torch.mv(A, x), A@x
```

### 2.3.11 Norms
Norm can be indicated with function: ||   || that will do: $\mathbb{R}^d \rightarrow \mathbb{R}$, where $d$ is the dimension of the vector. 

$\ell_1$: Is for L1 norms (a.k.a. Manhattan distance). By definition, it is the absolute values of the vector's elements: $ || \mathbf{x} || = \sum_i^n |x_i|$ 

$\ell_2$: is for L2 norms (a.k.a. Euclidean length of vector). By definition, it is: $||\mathbf{x}||_2 = \sqrt{\sum_i^n x_i^2}$

Note that both $\ell_1$ and $\ell_2$ norms are special cases of the more general $\ell_p$ norms: 
$$\Vert \mathbf{x} \Vert_p = \left( \sum_{i=1}^n |x_i|^p \right)^{1/p}$$

Frobenius norm where we take the $\ell_2$ norm of all elements in a $\mathbf{matrix}$. 

A norm is a function satisfies the following 3 properties:
1. Given any vector $\mathbf{x}$, if we scall all elements of the vector by a scalar $\alpha \in \mathbb{R}$, its norm will scale accordingly: $$ ||\alpha \mathbf{x}|| = |\alpha| ||\mathbf{x}|| $$
2. Any vectors $\mathbf{x}$ and $\mathbf{y}$, norms satisfy the following triangle inequality: $$ ||\mathbf{x+y}|| \le ||\mathbf{x}|| + ||\mathbf{y}|| $$
3. The norm of a vector is non-negative and it only vanishes if the vector is zero: $$ ||\mathbf{x}|| > 0, \text{for all } \mathbf{x}\neq 0$$

```python
# torch.norm to calculate the L2 norm of a vector
u = torch.tensor([3.0, -4.0])
torch.norm(u)
```

```python
# Frobenius norm of a matrix
torch.norm(torch.ones((4, 9)))
```

Results below show that (A @ B) @ C is takes less time overall than A @ (B @ C). This can be attributed to the fact that A @ B will result in 32,768 elements w.r.t. ~4 billion of B @ C, thus taking less memory and compute. Due to associativity of matrix manipulation, it will resultl in memory and compute time but the same result. 

Note, the B @ C operation took up ~17GB of memory w/ CPU at peak

```python
torch.cuda.is_available()
```

```python
import time

A = torch.randn(2**10, 2**16)  # (1024, 65536) = 67,108,864 elements
B = torch.randn(2**16, 2**5)   # (65536, 32)   = 2,097,152 elements
C = torch.randn(2**5, 2**16)   # (32, 65536)   = 2,097,152 elements

# A @ B = (1024, 65536) @ (65536, 32) = (1024, 32) = 32,768 elements
# B @ C = (65536, 32) @ (32, 65536) = (65536, 65536) = 4,294,967,296 elements
start1 = time.perf_counter()
temp1 = A @ B
ab_then_c = temp1 @ C
end1 = time.perf_counter()

print("Time for (A @ B) @ C: ", end1 - start1)

start2 = time.perf_counter()
temp2 = B @ C
bc_then_a = A @ temp2
end2 = time.perf_counter()

print("Time for A @ (B @ C): ", end2 - start2)
```

Exercise 11

Naively, both operations (A @ B and A @ C.T) should take roughly the same amount fo time and compute since they are of the same dimensions and all initialized with randn. A @ C.T takes slightly longer since it does not copy the data, but tells pyrotch to interpret same memory with swapped dimensions. 

However, you can see below that the latter (A @ (C=B.T).T ) takes slightly longer than the previous 2 due to the slightly increased overhead. 

```python
import time
import torch

A = torch.randn(2**10, 2**16)  # (1024, 65536) = 67,108,864 elements
B = torch.randn(2**16, 2**5)   # (65536, 32)   = 2,097,152 elements
C = torch.randn(2**5, 2**16)   # (32, 65536)   = 2,097,152 elements

start1 = time.perf_counter()
AB = A @ B
end1 = time.perf_counter()
print("Time for A @ B: ", end1 - start1)

start2 = time.perf_counter()
AC_T = A @ C.T
end2 = time.perf_counter()
print("Time for A @ C.T: ", end2 - start2)

# Try initializing C = B.T without cloning memory
C = B.T

print("C contiguous:", C.is_contiguous())
print("C.T contiguous:", C.T.is_contiguous())

start3 = time.perf_counter()
AC_T_view = A @ C.T 
end3 = time.perf_counter()
print("Time for A @ C.T where C = B.T:", end3 - start3)
```
